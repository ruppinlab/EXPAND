```python
#### ------------------------------------------------------------------------------------------
#### author: Ranjan Barman, date: Jun 2, 2025
#### Mapped CPTAC HoverNet Immune-Only NPIFs (all tiles) to TCGA_BRCA subtype status
#### --------------------------------------------------------------------------------------------

import os
import pandas as pd

# Set working directory
_wpath_ = "/data/Lab_ruppin/Ranjan/HnE/"
os.chdir(_wpath_)
print(f"Working directory: {_wpath_}\n")

# Define dataset name and file paths
dataset_name = "CPTAC_BRCA"

# File paths for immune-only NPIFs and clinical subtypes
CPTAC_BRCA_ImmuneOnly_NPIFs_file = f"{dataset_name}/HoverNet/outputs/CPTAC_BRCA_HoverNet_ImmuneOnly_NPIFs.csv"
CPTAC_subtypes_file = "/data/Lab_ruppin/dhrubas2/HnE/CPTAC_BRCA/processed/CPTAC_BRCA_clinical_data_summary_matched.tsv"

# Load the immune-only NPIFs data
CPTAC_BRCA_ImmuneOnly_NPIFs = pd.read_csv(CPTAC_BRCA_ImmuneOnly_NPIFs_file)
CPTAC_BRCA_ImmuneOnly_NPIFs.columns = CPTAC_BRCA_ImmuneOnly_NPIFs.columns.str.strip()

# Load the clinical subtypes data
CPTAC_subtypes_data = pd.read_table(CPTAC_subtypes_file, sep="\t")
CPTAC_subtypes_data = CPTAC_subtypes_data[["Patient_ID", "HER2_status", "PR_status", "ER_status"]]
CPTAC_subtypes_data = CPTAC_subtypes_data.rename(columns={
    "HER2_status": "HER2_Status",
    "PR_status": "PR_Status",
    "ER_status": "ER_Status"
})

# Convert and pad patient IDs
CPTAC_subtypes_data["Patient_ID"] = CPTAC_subtypes_data["Patient_ID"].astype(str).str.zfill(3)

CPTAC_BRCA_ImmuneOnly_NPIFs
CPTAC_subtypes_data


```

```python
# Print debug info
print("Patient_IDs from CPTAC_BRCA_HoverNet_ImmuneOnly_NPIFs (Total: {}):".format(len(CPTAC_BRCA_ImmuneOnly_NPIFs)))
print(CPTAC_BRCA_ImmuneOnly_NPIFs["Patient_ID"].unique())

print("\nPatient IDs from CPTAC_subtypes_data (Total: {}):".format(len(CPTAC_subtypes_data)))
print(CPTAC_subtypes_data["Patient_ID"].unique())

# Identify non-matching IDs
slide_names = set(CPTAC_BRCA_ImmuneOnly_NPIFs["Patient_ID"])
patient_ids = set(CPTAC_subtypes_data["Patient_ID"])
non_matching_slides = slide_names - patient_ids
non_matching_patients = patient_ids - slide_names

print("\nImmune-Only NPIFs that do NOT have a matching Patient_ID:")
print(non_matching_slides)
print("\nBRCA subtypes that do NOT have a matching Patient_ID:")
print(non_matching_patients)
```

```python
# Merge data
merged_df = pd.merge(CPTAC_subtypes_data, CPTAC_BRCA_ImmuneOnly_NPIFs, on='Patient_ID', how="inner")

# Save output
output_path = f"{dataset_name}/outputs/HoverNet/Subtypes/"
os.makedirs(output_path, exist_ok=True)
output_file = os.path.join(output_path, "HoverNet_ImmuneOnly_NPIFs_All_Tiles_CPTAC_BRCA_Mapped_BRCA_Status.csv")
merged_df.to_csv(output_file, index=False)

print(f"Mapped immune-only NPIF data saved to: {output_file}")
print("Done!")

merged_df
```
