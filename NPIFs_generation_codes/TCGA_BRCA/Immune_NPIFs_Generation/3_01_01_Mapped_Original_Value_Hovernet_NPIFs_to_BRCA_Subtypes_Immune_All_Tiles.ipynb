```python
#### ------------------------------------------------------------------------------------------
#### author: Ranjan Barman, date: May 30, 2025 (modified for immune only)
#### Mapped HoverNet 12 immune-only NPIFs to TCGA_BRCA subtype status (HER2, PR, ER)
#### ------------------------------------------------------------------------------------------

import os
import pandas as pd

# Set working directory
_wpath_ = "/data/Lab_ruppin/Ranjan/HnE/"
os.chdir(_wpath_)
print(f"Working directory: {_wpath_}\n")

# Dataset and input paths
dataset_name = "TCGA_BRCA_FFPE"

# File paths
immune_npifs_file = f"{dataset_name}/outputs/HoverNet/HoverNet_NPIFs_TCGA_BRCA_1106_AllTiles_ImmuneOnly.csv"
tcga_subtypes_file = "PA_HIF_BRCA/outputs_clinical/TCGA_BRCA_Subtypes_clinical.tsv"

# Load immune-only NPIFs
immune_npifs_df = pd.read_csv(immune_npifs_file)

# Ensure Slide_Name is in consistent format
immune_npifs_df["Slide_Name"] = immune_npifs_df["Slide_Name"].str[:12]
immune_npifs_df = immune_npifs_df.drop_duplicates(subset=["Slide_Name"], keep="first")

# Load subtype data
tcga_subtypes_data = pd.read_table(tcga_subtypes_file, sep="\t")

# Keep and rename subtype columns
tcga_subtypes_data = tcga_subtypes_data[["sampleID", "HER2_Final_Status_nature2012", "PR_Status_nature2012", "ER_Status_nature2012"]]
tcga_subtypes_data = tcga_subtypes_data.rename(columns={
    "HER2_Final_Status_nature2012": "HER2_Status",
    "PR_Status_nature2012": "PR_Status",
    "ER_Status_nature2012": "ER_Status"
})

# Standardize sample ID by removing '-01' suffix
tcga_subtypes_data["sampleID"] = tcga_subtypes_data["sampleID"].str.replace("-01", "", regex=False)

tcga_subtypes_data

```

```python
# Merge on sample ID and Slide_Name
merged_df = pd.merge(tcga_subtypes_data, immune_npifs_df, left_on="sampleID", right_on="Slide_Name", how="inner")
merged_df.drop(columns=["Slide_Name"], inplace=True)

# Output path and file name
output_path = f"{dataset_name}/outputs/HoverNet/Subtypes/"
os.makedirs(output_path, exist_ok=True)
output_file = os.path.join(output_path, "HoverNet_ImmuneOnly_NPIFs_Mapped_BRCA_Status_All_tiles.csv")

# Save merged data
merged_df.to_csv(output_file, index=False)

print(f"Mapped immune-only NPIFs data saved to: {output_file}")
print("Done!")

# Return merged DataFrame if running interactively
merged_df
```
